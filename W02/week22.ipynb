## Mission 4：公式計算與表現指標

**Knowledge Point**: 算術運算子與學習表現公式計算。

**計算公式**：
$$ \text{預測小測驗分數} = 50 + 3 \times \text{讀書時數} + 2 \times \text{睡眠時數} - 1.2 \times \text{螢幕使用時數} + 0.15 \times \text{出席率} $$

```python
# 單一學生特徵資料
study_hours = 2.5
sleep_hours = 7.0
screen_time_hours = 4.0
attendance_rate = 90

# 計算預測表現分數
predicted_score = (
    50
    + 3 * study_hours
    + 2 * sleep_hours
    - 1.2 * screen_time_hours
    + 0.15 * attendance_rate
)

print("計算所得的預測分數:", predicted_score)
```

## Mission 5：自訂函數與條件判斷

**Knowledge Point**: 使用 `def` 定義函數與 `if / elif / else` 邏輯判斷。

```python
def describe_score(score):
    if score >= 80:
        return '表現優秀'
    elif score >= 60:
        return '達基本標準'
    else:
        return '需要加強'

# 測試函數
print("85 分評價:", describe_score(85))
print("70 分評價:", describe_score(70))
print("50 分評價:", describe_score(50))
```

## Mission 6：載入套件與設定隨機種子

**Knowledge Point**: `import` 套件、別名 (`as`)、`NumPy` 隨機生成器設定。

```python
import numpy as np
import pandas as pd

# 設定隨機種子以確保結果可重現
random_seed = 42
rng = np.random.default_rng(random_seed)
print("套件載入成功，隨機種子已設定！")
```

--- 
## Mission 7：建立真實的大學生學習資料集

**Knowledge Point**: 使用 `pandas DataFrame` 生成 60 筆模擬學生資料。

```python
student_count = 60
n = student_count

# 1. 學號 (1~60)
student_ids = np.arange(1, n + 1)

# 2. 系所（隨機抽樣）
depts = rng.choice(["資訊管理", "企業管理", "財務金融"], size=n)

# 3. 睡眠時數 (常態分佈 Mean=6.7, SD=1.1, 四捨五入至小數一位)
sleeps = np.round(rng.normal(loc=6.7, scale=1.1, size=n), 1)

# 4. 螢幕使用時數 (Mean=5.0, SD=1.5, 限制在 1~10 小時)
screens = np.round(np.clip(rng.normal(loc=5.0, scale=1.5, size=n), 1.0, 10.0), 1)

# 5. 讀書時數 (Mean=2.5, SD=1.0, 限制在 0.5~6 小時)
studies = np.round(np.clip(rng.normal(loc=2.5, scale=1.0, size=n), 0.5, 6.0), 1)

# 6. 出席率 (Mean=88, SD=8, 限制在 60~100)
attendances = np.round(np.clip(rng.normal(loc=88, scale=8, size=n), 60, 100), 0)

# 建立 DataFrame
df = pd.DataFrame({
    "student_id": student_ids,
    "department": depts,
    "sleep_hours": sleeps,
    "screen_time_hours": screens,
    "study_hours": studies,
    "attendance_rate": attendances
})

# 顯示前幾筆資料
df.head()
```

--- 
## Mission 8：建立綜合指標 (quiz_score)

**Knowledge Point**: 將公式應用於整個 DataFrame 欄位，並加上隨機殘差項。

```python
# 加上殘差項 (平均 0, 標準差 5)
noise = rng.normal(loc=0, scale=5, size=n)

# 依公式計算小測驗分數
raw_quiz = (
    50
    + 3 * df["study_hours"]
    + 2 * df["sleep_hours"]
    - 1.2 * df["screen_time_hours"]
    + 0.15 * df["attendance_rate"]
    + noise
)

# 限制分數在 0~100 之間並轉為整數
df["quiz_score"] = np.clip(raw_quiz, 0, 100).round().astype(int)

# 檢視前 5 筆資料欄位
df[["student_id", "study_hours", "sleep_hours", "screen_time_hours", "attendance_rate", "quiz_score"]].head()
```

## Mission 9：加入缺失值與異常值

**Knowledge Point**: 模擬真實資料中的缺值 (`np.nan`) 與極端值，並使用 `.loc` 進行特定位置的修改。

```python
# 注入缺失值 (NaN)
df.loc[[5, 17], "sleep_hours"] = np.nan

# 注入極端異常值
df.loc[8, "screen_time_hours"] = 15.0

print("已成功插入缺值與極端異常值！")
```

## Mission 11：條件篩選資料

**Knowledge Point**: 使用布林條件篩選資料列。

```python
# 篩選出測驗分數大於等於 80 分的學生
high_score_condition = df["quiz_score"] >= 80
high_score_students = df[high_score_condition]

print(f"高分群學生共有 {len(high_score_students)} 人：")
print(high_score_students.head())
```

**Knowledge Point**: 將整理好的 DataFrame 匯出為 CSV 檔，並提供 Google Colab 下載指令。
